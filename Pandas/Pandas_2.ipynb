```python
#Handling null values in a csv file
#Instead of removing complete row having null values we will fill those null values 
#as by removing we may lost some important data
df=pd.read_csv(r'C:\Users\hp\Downloads\sample.csv')
df.head(10)
```

```python
df.isnull().sum()
```

```python
#Fill null or NaN values
#Filling null values with zero
df1=df.fillna(0)
df1.head(10)
```

```python
df1=df.fillna(2)
df1.head(10)
```

```python
#Here each column want to fill null values in there col with different values
#like here physics want to fill Nan value with none, Maths want to fill Nan value with 30
df1=df.fillna({"Physics":"none","Chemistry":0,"Maths":30,"Computer":"-1"})
df1
```

```python
#Forward fill
#Forward fill - fills the null values with just the above field value in same column
df.fillna(method="ffill") #row wise fill i.e field value just in above row here default axis=0
```

```python
#Forward fill col wise
df.fillna(method="ffill",axis=1) #column wise fill i.e field value just previous in before column
```

```python
#Fill null values with some particular calculated value
#see 8, 10, 14 index which have null values
df["Physics"].fillna(value=df["Physics"].mean())
```

```python
#Backward fill - same as forward fill just the direction of taking values is opposite 
#insted of previous , just after field value is taken
#axis=0 below value of just below row (default)
df.fillna(method="bfill")
```

```python
#axis=1 side after value of just after column
df.fillna(method="bfill",axis=1)
```

```python
df.fillna(method="bfill",inplace=True)
df
```

```python
#sometimes we cannot use axis inplace and method at the same time
#this is due to Nan type 
#best is to change all Nan with np.nan first And then due bfill,inplace,axis together
dx=pd.read_csv(r'C:\Users\hp\Downloads\sample.csv')
print(dx.head(15))
```

```python
#First changing NaN values to np.nan value
dx.fillna({"Physics":np.nan,"Chemistry":np.nan,"Maths":np.nan,"Computer":np.nan},inplace=True)
print(dx)
dx.fillna(method="bfill",inplace=True)
print("After filling and saving changes in same dataframe:\n",dx.isnull().sum())
```

```python
#Replacing any value
df=pd.read_csv(r'C:\Users\hp\Downloads\sample.csv')
df.head()
```

```python
#all places where 26 is there is replaced by 30
df1=df.replace(to_replace=26,value=30)
df1.head()
```

```python
#another way - directly write values
df1=df.replace(27,100)  #first parameter is the value to be replaced and second one is the value which replacing old one
df1.head() 
```

```python
#replacing multiple values by the same single value
print("Printing Before replacing : (top 5 rows):\n",df.head())
df1=df.replace(to_replace=[56,26,27,28],value='A')
print("\nPrinting After replacing : (top 5 rows):\n",df1.head())
```

```python
#replacing multiple values with corresponding multiple values
df1=df.replace(to_replace=[26,27,28,16,17],value=['a','b','c','c','d'])
df1.head(15)
```

```python
#changes made only to physics column and saved in original dataframe due to inplace=True
df["Physics"].replace(to_replace=[12,16,22,90,20],value=['a','b','c','c','d'],inplace=True)
df.head(15)
```

```python
#replacing using regex
df1=df.replace('[A-Za-z]',0.0,regex=True)
df1.head(15)
```

```python
#replacing by forward and backward fill
df1=df.replace(to_replace=[26,27,28],method="bfill")
df1.head(8)
```
