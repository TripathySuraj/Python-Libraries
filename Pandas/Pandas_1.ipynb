```python
#Creating a csv file and populating data in it
import csv

data = [
    [1.0, 7890.0],
    [2.0, 6000.8],
    [3.0, 8500.5],
    [8.0, 12000.0],
    [2.0, 6100.0],
    [9.0, 13500.2],
    [4.5, 9000.7],
    [5.0, 9700.0],
    [6.0, 10500.3],
    [7.5, 11200.6],
    [3.2, 8200.4],
    [1.5, 7300.1],
    [2.8, 7900.0],
    [4.0, 8800.0],
    [6.5, 10800.0],
]

#newline=''	This prevents Python from inserting extra blank lines between rows when writing CSV files on Windows. 
#It ensures newlines are handled consistently.
with open("Salary_Data.csv",'w',newline='') as file:    #if Salary_Data.csv is not there it will create it
    writer=csv.writer(file)
    writer.writerow(["YearsExperience", "Salary"])
    writer.writerows(data)

#Read file content
with open("Salary_Data.csv","r") as file:
    contents=file.read()
print("Files in Salary.csv:\n",contents)
```

```python
#Handling Null datas
df3=pd.read_csv(r'C:\Users\hp\Downloads\sample.csv')
print("Top 5 rows:\n",df3.head())
df3.isnull() #gives true for null value
```

```python
#Getting number of null values
print("Getting sum of null cells or values for each column :\n",df3.isnull().sum())
print("Total of all null values:",df3.isnull().sum().sum())
```

```python
#Dropping rows consisting of null values or NaN values
#axis=1 for col and axis=0 for rows

#All rows consisting of atleast one null value is dropped
print("Before dropping rows with a null value (rows,cols):",df3.shape)

#Dropping rows with atleast a null value
df4=df3.dropna() #default axis = 0 (row wise) , in numpy axis=1 is for row wise sum in matrix

print("After dropping rows with a null value (rows,cols):",df4.shape)
```

```python
#Dropping columns consisting of null values or NaN values
#axis=1 for col and axis=0 for rows

#All columns consisting of atleast one null value is dropped
print("Before dropping rows with a null value (rows,cols):",df3.shape)

#Dropping columns with atleast a null value and storing in another dataframe
df4=df3.dropna(axis=1)  #axis=1 for column wise , in numpy axis=0 is for column wise sum in matrix

print("After dropping rows with a null value (rows,cols):",df4.shape)
```

```python
#Dropping rows accoring to any or all
print("Shape Before dropping rows:",df3.shape)
dfx=df3.dropna(how="any") #if any row value is null remove that row
print("After dropping rows with any null value:",dfx.shape)

dfy=df3.dropna(how="all") #if all row values is null remove that row
print("After dropping rows with all null values:",dfy.shape)
```

```python
#replacing original datafram i.e changes will be done in original dataframe
#this is done by setting inplace=True
print("Shape Before dropping cols :",df3.shape)
df3.dropna(axis=1,inplace=True,how='any')
print("Shape after dropping cols :",df3.shape)
```

```python
#accessing different column in csv file
df=pd.read_csv("Salary_Data.csv")
print("Accessing Salary column data:\n",df["Salary"])
```

```python
#Filtering rows
df1=df[(df["YearsExperience"]>4) & (df["Salary"]>8000)]
print("Conditional statement with 'and' operator:\n",df1)

df2=df[(df["YearsExperience"]>7) | (df["Salary"]>10000)]
print("\nConditional statement with 'or' operator:\n",df2)

df3=df[~(df["YearsExperience"]>7)]
print("\nConditional statement with 'not' operator:\n",df3)
```
