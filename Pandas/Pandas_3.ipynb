```python
#loc() and iloc()
#difference is loc consider index according to given one while in iloc index always goes as 0,1,2,3,4,5......
#like here below Roll no is set as index so idex will be according to Roll No. values so it will like 1,2,3,4,... in loc
#while in iloc it will still be as 0,1,2,3,....
#in short in loc index is equal to Name of row
#loc and iloc can have different functionalities or different ways of accessing rows, columns or data
#to access one more way is df[col Name][row Name]
```

```python
#Merge Function - merge two dataframes according to an intersecting or common column
dfx=pd.DataFrame({"Roll No.":[1,2,3,4,5],"Maths":[34,56,88,21,97]})
dfx
```

```python
dfy=pd.DataFrame({"Roll No.":[1,2,3,4,5],"Physics":[29,47,88,72,67]})
dfy
```

```python
pd.merge(dfx,dfy,on="Roll No.")
```

```python
pd.merge(dfy,dfx,on="Roll No.")
```

```python
pd.merge(dfx,dfy) #on default intersecting column which is Roll No. is taken
```

```python
#When the intersecting column has different values
#here only the  values of intersecting col that is in both DataFrame will produce the output
df1=pd.DataFrame({"Roll No.":[1,2,3,4,5],"Maths":[34,56,88,21,97]})
df2=pd.DataFrame({"Roll No.":[1,2,7,4,9],"Physics":[29,47,88,72,67]})
pd.merge(df1,df2) #here Roll no 1,2 & 4 is present in both so they will be there in output after merging
```

```python
#to get remaining unmatched rollno. in above case
#according to any one dataframe
pd.merge(df1,df2,how="left")
```

```python
pd.merge(df1,df2,how="right")
```

```python
#outer will take all the rollno.
pd.merge(df1,df2,how="outer")
```

```python
#append() function - appends one dataframe to another
df1=pd.DataFrame({"Roll No.":[1,2,3,4,5],"Maths":[34,56,88,21,97],"Physics":[29,47,88,72,67]})
df2=pd.DataFrame({"Roll No.":[6,7,8,9,10],"Maths":[84,59,61,85,91],"Physics":[82,57,81,44,60]})
print(df1)
print(df2)
```

```python
#Pivot table
dfp=pd.read_csv(r'C:\Users\hp\Downloads\sample2.csv')
dfp.head()
```

```python
#here data will be mean of all marks belonging to the particular Branch and Subject
#like (CS,Chemistry) - 30.888889 here it means mean of all marks belonging to Cs branch and chemistry subject
#like (CS,Roll No.) - 16 here it means mean of all roll nos. belonging to cs branch
#so we can find out by seeing average marks of cs students in chemistry subject, in Maths subject,etc respectively
#while calculating if 11 rows are there out of which 2 are null or NaN then calculation will be (marks in those 9 cols/9)
pd.pivot_table(dfp,index="Branch")
```

```python
#by default pivot calculates mean or average according to index
#we can also calculate other parameters by using aggfunc
#like here we are calculating sum according to index instead of mean
pd.pivot_table(dfp,index="Branch",aggfunc="sum")
```

```python
x=dfp["Chemistry"][dfp["Branch"]=="CS"]
print(x)
print(len(x))
```

```python
pd.pivot_table(dfp,index="Branch",aggfunc="count")   #here NaN will be excluded
```

```python
pd.pivot_table(dfp,index="Branch",aggfunc="max")
```

```python
#calculation is shown according to data belonging to given branch in given section like mix marks in (CS,A)
pd.pivot_table(dfp,index="Branch",columns="Section",aggfunc="min")
```

```python
pd.pivot_table(dfp,index="Branch",columns="Section")
```

```python
#dropping duplicate values
df=pd.DataFrame({
    "Rollno":pd.Series(np.arange(1,6)),
    "Name":pd.Series(["Rahul","Priyanka","Bablu","Rahul","Shyam"]),
    "Age":pd.Series(np.random.randint(10,20,5))
})
df
```

```python
df.drop_duplicates(subset=["Name"])   #by default first pone is kept rest cols are dropped
```

```python
#using keep which decides whi occurences will remain intact after dropping duplicates
df.drop_duplicates(subset=["Name"],keep="first")  #first Rahul in index 1 is kept
```

```python
df.drop_duplicates(subset=["Name"],keep="last")  #Last Rahul in index 4 is kept rest are dropped here i.e at index 1 is dropped
```

```python
df.drop_duplicates(subset=["Name"],keep=False) #all rows with duplicate values in the specified column are dropped
```

```python
df["Name"].value_counts(dropna=True)  #giving counts of each unique data in 'Name' column excluding NaN values
```
