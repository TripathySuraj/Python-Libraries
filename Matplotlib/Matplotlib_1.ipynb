```python
temperature_pune=list(np.random.randint(5,46,10))
humidity_pune=list(np.random.randint(18,60,10))

temperature_banglore=list(np.random.randint(20,36,10))
humidity_banglore=list(np.random.randint(34,48,10))

humidity_banglore
```

```python
#Bar Plot
subjects=["Maths","English","Science","Social Studies","Computer"]
marks=[89,90,45,78,99]

plt.bar(subjects,marks)
plt.show()
```

```python
plt.bar(subjects,marks,color="green")
plt.show()
```

```python
#giving different colours to difference bars and giving width to the bars
colors=["red","blue","yellow","green","pink"]
plt.bar(subjects,marks,color=colors,width=0.6)  #width - 0.6 meeans only 60 % of original width will be taken
plt.show()
```

```python
#giving border by "edgecolor" to bars and linewidth i.e border width
plt.bar(subjects,marks,color=colors,width=0.6,edgecolor="white",linewidth=4)  #width - 0.6 meeans only 60 % of original width will be taken
plt.show()
```

```python
#changing style of the border line by "linestyle"
plt.bar(subjects,marks,color=colors,width=0.6,edgecolor="white",linewidth=4,linestyle="--") #line style can be :,-.,-,--
plt.show()
```

```python
#Drawing Bar plot in horizontal direction
plt.barh(subjects,marks,color=colors)   #here bar direction is horizontal
plt.ylabel("Subjects")
plt.xlabel("Marks")
plt.show()
```

```python
plt.barh(subjects,marks,color=colors,height=0.6,edgecolor="white",linewidth=4,linestyle="-")   #here bar direction is horizontal
plt.ylabel("Subjects") #here instead of width, height is used to change the thickness of bars
plt.xlabel("Marks")
plt.show()
```

```python
subjects=["Maths","Physics","Science","Social Studies","Computer"]
marks1=np.random.randint(30,100,5)
marks2=np.random.randint(15,92,5)
print("Marks 1:",marks1)
print("Marks 2:",marks2)
```

```python
#drawing two Bar Plots on same graph
plt.figure(figsize=(8,8))

plt.bar(subjects,marks1,width=0.6,color="yellow")
plt.bar(subjects,marks2,width=0.6,color="red")

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()

print("Here one bar covers the other if it is bigger in value hence other one is not visible")

#here the problem is one bar can cover the other one completely if it has bigger value
```

```python
#Solution to above case
subjects_len=np.arange(len(subjects))  #same as np.arange(0,len(subjects))
print(subjects_len)
width=0.4
```

```python
colors=["red","blue","yellow","green","pink"]
plt.figure(figsize=(8,8))

plt.bar(subjects_len,marks1,width=width,color=colors)
plt.bar(subjects_len+width,marks2,width=width,color=colors,alpha=0.4)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```
