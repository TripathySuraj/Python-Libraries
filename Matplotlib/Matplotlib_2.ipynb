```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
plt.style.use("dark_background")
```

```python
#Histogram
#Histogram is used to depict range of data related graph
#like how many students scored in range 10-20 marks, 20-30 marks,etc
marks_50_students=np.random.randint(5,100,(50))
marks_50_students
```

```python
#rwidth is the width of the bars
plt.figure(figsize=(8,8))
plt.xticks(np.arange(0,100,10))
plt.yticks(np.arange(0,12,2))

plt.hist(marks_50_students,edgecolor="yellow",color="red",linewidth=4)

plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.show()
```

```python
#in the above figure you can see ranges are not conciding with the mentioned range values on x-axis 
#for this bins are created
#data between the bin ranges will be there
#First define bins then set bins in plt.hist() and then define xticks with same range as bins

bins=np.arange(0,100,10)

plt.figure(figsize=(8,8))
plt.hist(marks_50_students,bins=bins,edgecolor="yellow",color="green",linewidth=2)

plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.xticks(np.arange(0,100,10))
plt.show()
```

```python
#to rotate the graph
bins=np.arange(0,100,10)

plt.figure(figsize=(8,8))
plt.hist(marks_50_students,bins=bins,edgecolor="yellow",color="blue",linewidth=2,orientation="horizontal")

plt.xlabel("Number of students")
plt.ylabel("Marks")
plt.yticks(np.arange(0,100,10))
plt.show()
```

```python
#rwidth defines the width of the bar 
#this helps to visualize values in certain range more clearly
bins=np.arange(0,100,10)

plt.figure(figsize=(8,8))
plt.hist(marks_50_students,bins=bins,edgecolor="yellow",color="blue",linewidth=2,rwidth=0.6)

plt.ylabel("Number of students")
plt.xlabel("Marks")
plt.xticks(np.arange(0,100,10))
plt.show()
```

```python
#Types of histogram
```

```python
#Hist types - 'bar','barstacked','step','stepfilled'
bins=np.arange(0,100,10)

plt.figure(figsize=(8,8))
plt.hist(marks_50_students,bins=bins,color="blue",histtype="step")

plt.ylabel("Number of students")
plt.xlabel("Marks")
plt.xticks(np.arange(0,100,10))
plt.show()
```

```python
#Two hist plots in same graph

marks_50_student1=np.random.randint(0,100,50)
marks_50_student2=np.random.randint(0,100,50)
print(marks_50_student1)
print(marks_50_student2)
```

```python
bins=np.arange(0,100,5)

plt.figure(figsize=(8,8))
plt.hist([marks_50_student1,marks_50_student2],bins=bins,color=["blue","yellow"])
#bins can also be given in lists
plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.title("Marks of students of 2 Classes")
plt.xticks(np.arange(0,100,5))
plt.show()

#from the graph we can say marks in range 75-80 students from class 1 is in more numbers
```

```python
#Pie Plot
classes=["Physics","Chemistry","Maths","Science","SST"]
marks=[45,89,92,78,33]
```

```python
plt.pie(marks,labels=classes)
plt.show()
```

```python
colors=["red","purple","yellow","orange","green"]
plt.pie(marks,labels=classes,colors=colors)
plt.show()
```

```python
#to show percentage of each part in pie chart
#this is done by autopct
colors=["red","purple","yellow","orange","green"]
plt.pie(marks,labels=classes,colors=colors,autopct="%0.2f%%") 
#0.2f means 2 decimal points 0.1f will give one decimal no. like 9.8
plt.show()
```

```python
#explode hepls to make any part exploded i.e take that part slightly out of the pie
#explode values can be from 0 to 1 as value is increased it will be exploaded more
#here chemistry is exploded if 0.1 is changed to 0.2 it will be exploded out more
explode_values=[0,0.1,0,0,0]
colors=["red","purple","yellow","orange","green"]
plt.pie(marks,labels=classes,colors=colors,autopct="%0.2f%%",explode=explode_values) 
plt.show()
```

```python
#by using shadow you can see some shadow kind of thing in slices
plt.style.use("default")
explode_values=[0,0.1,0,0,0]
colors=["red","purple","yellow","orange","green"]
plt.pie(marks,labels=classes,colors=colors,autopct="%0.2f%%",explode=explode_values,shadow=True) 
plt.show()
```

```python
plt.style.use("dark_background")
```

```python
#radius help to adjust size of pie chart
explode_values=[0.1,0.1,0,0,0]
colors=["red","purple","yellow","orange","green"]
plt.pie(marks,labels=classes,colors=colors,autopct="%0.2f%%",explode=explode_values,radius=1.3) 
#plt.title("Subject pie chart")
plt.show()
```

```python
#textprops is used to change the size and color of texts in pie chart
textprops={"fontsize":14,"color":'brown'}
plt.figure(figsize=(8,8))
explode_values=[0.1,0.2,0,0,0]
colors=["red","purple","yellow","orange","green"]
plt.pie(marks,labels=classes,colors=colors,autopct="%0.2f%%",explode=explode_values,textprops=textprops)
plt.show()
```

```python
#wedgeprops changes the properties of each wedge which means each slice here indivisually is a wedge
#here wedgeprops gives border and changes its style,color and width
textprops={"fontsize":14,"color":"white"}
wedgeprops={"linewidth":3,"linestyle":'--',"edgecolor":"white"}

plt.pie(marks,labels=classes,colors=colors,autopct="%0.2f%%",explode=[0,0.1,0,0.2,0],textprops=textprops,wedgeprops=wedgeprops)
plt.show()
```

```python
#legend is for side chart which shows which color is for which classes
textprops={"fontsize":14,"color":"white"}
wedgeprops={"linewidth":3,"linestyle":'-',"edgecolor":"white"}
plt.figure(figsize=(8,8))
plt.pie(marks,labels=classes,colors=colors,autopct="%0.2f%%",explode=[0,0,0,0.1,0],textprops=textprops,wedgeprops=wedgeprops)
plt.legend()
plt.title("Subjects and Average scores",color="blue",fontsize=20)
plt.show()
```

```python
study_hours=np.sort(np.random.randint(2,13,15))
marks=np.random.randint(5,100,15)
print(study_hours)
print(marks)
```

```python
#scatterplot
plt.figure(figsize=(8,8))
plt.plot(study_hours,marks,"y*",markersize=15)
plt.show()
```

```python
#line graph
plt.plot(study_hours,marks,color="blue",linestyle="--",linewidth=3)
plt.show()
```

```python
#histogram
plt.hist(marks)
plt.show()
```

```python
#scatterplot and line plot in same graph
plt.figure(figsize=(6,6))
plt.plot(study_hours,marks,"r*",markersize=20)
plt.plot(study_hours,marks,color="blue",linestyle="--",linewidth=3)
plt.show()
```

```python
#subplots - making all graphs come into same picture

plt.figure(figsize=(10,10))  #here this is very much necessary
#subplot can be viewd as a square divided in four equal parts in this case so there are two rows and two cols
#and four indices i.e index 1,2,3 and 4 - one index for each part 
#first row consists of index 1 and 2 in column 1 and 2 respectively
#2nd row consists of index 3 and 4 in column 1 and 2 respectively

#sactterplot in index 1
plt.subplot(2,2,1) #(rows,cols,index no.) - here rows is 2, cols is 2 and index is 1
plt.plot(study_hours,marks,"r*",markersize=15)

#line plot in index 2
plt.subplot(2,2,2) #(rows,cols,index no.) - here rows is 2, cols is 2 and index is 2
plt.plot(study_hours,marks,color="blue",linestyle="--",linewidth=3)

#Hist plot in index 3
plt.subplot(2,2,3) #(rows,cols,index no.) - here rows is 2, cols is 2 and index is 3
bins=np.arange(0,100,10)
plt.hist(marks,bins=bins,color="green",edgecolor="white",linewidth=4)
plt.xticks(np.arange(0,100,10))

#Comination of Scatter-Line plot in index 4
plt.subplot(2,2,4) #(rows,cols,index no.) - here rows is 2, cols is 2 and index is 4
plt.plot(study_hours,marks,"r*",markersize=15)
plt.plot(study_hours,marks,color="blue",linestyle="--",linewidth=3)

#Saving plot in desired location by savefig
#first method of passing file path - backward slash with raw path
plt.savefig("C:/Users/hp/Downloads/subplot.png", quality=100, facecolor='b') #for blue background  

#second method of passing file path - by raw path \ forward slash
plt.savefig(r'C:\Users\hp\Downloads\subplot2.png', quality=100, facecolor='k') #for black background

plt.show()
```
