```python
#Classes
class Person:
    def __init__(self,name,age):  #self is used to signify that current object is used
        self.name=name
        self.age=age
        
    def getInfo(self):   #self puts function into action for current object only
        n="%s" % self.age #convert int to string
        str="Name : "+self.name+", Age : "+n
        #print("Hello")
        return str
        
person1=Person("John",30)
str=person1.getInfo()
print(str)


person2=Person("Ramesh",56)
str1=person2.getInfo()
print(str1)


#Polymorphism
print("\nPolymorphism")
class Animal:
    def __init__(self,species):
        self.species=species
    
    def makeSound(self):
        print("Unknown sound...")
        
class Cat(Animal):
    def makeSound(self):
        print("I am a",self.species,".I make sound of Meow...")
        
class Dog(Animal):
    def makeSound(self):
        print("I am a",self.species,".I make sound of Bark...")
        
dog=Dog("dog")
#dog.makeSound()
cat=Cat("cat")
#cat.makeSound()
animal=Animal("Lion")
#animal.makeSound

for i in (dog,cat,animal):
    i.makeSound()
    
#Inheritance
class Vehicle:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
        
    def startEngine(self):
        print("Engine of",self.model,"is starting...")
        

class Car(Vehicle):
    def __init__(self,make,model,year,numberOfWheels):
        super().__init__(make,model,year)
        self.numberOfWheels=numberOfWheels
        
    def drive(self):
        print("Driving the car...")
        
vehicle=Vehicle("Toyota","Corolla",2024)
car=Car("Tesla","Model S",2023,5)

vehicle.startEngine()
car.startEngine() #-child inherit properties from parent
car.drive()
#vehicle.drive() - parent cannot inherit properties from child
```

```python
#programs
#3)print nos. from 1 to 100 but for multiple of 3 print Fizz
#for multiple of 5 print Buzz and for multiples of both 5 & 3 print FizzBuzz

def multipleOfThree(num):
    sum=0
    while num>0:
        rem=num%10
        sum=sum+rem
        num=num//10  #floor division
    if sum%3==0:
        return True
    else:
        return False
    
def multipleOfThree1(num):
    if num%3==0:
        return True
    else:
        return False 

def multipleOfFive(num):
    if num%10==0 or num%10==5:
        return True
    else:
        return False
    
for i in range(1,101):
    if(multipleOfThree(i) and multipleOfFive(i)):
        print("FizzBuzz",end=' ')
    elif(multipleOfThree(i)):
        print("Fizz",end=' ')
    elif(multipleOfFive(i)):
        print("Buzz",end=' ')
    else:
        print(i,end=' ')
     
```
