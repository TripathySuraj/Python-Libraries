```python
#Function to Replace N-th Occurrence of a Substring

def replace_nth(text, sub, repl, n):
    """Replaces the nth occurrence of `sub` in `text` with `repl`."""
    start = -1
    for i in range(n):
        start = text.find(sub, start + 1)
        if start == -1:
            return text  # Not enough occurrences

    return text[:start] + repl + text[start + len(sub):]

s = "apple banana apple orange apple"
result = replace_nth(s, "apple", "mango", 2)
print(result)
```

```python
#Greatest common divisor
def GCD(n1,n2):
    if n1==n2:
        return n1
    
    nx=lambda n1,n2:n1 if n1<n2 else n2
    ny=lambda n1,n2:n1 if n1>n2 else n2
    
    nlt=int(nx(n1,n2))
    ngt=int(ny(n1,n2))
    dic1={}
    dic1[nx]=1
    for i in range(2,nlt):
        if dic1.get(i,0)==0:
            if nlt%i==0:
                x=nlt/i
                dic1[i]=1
                dic1[nlt]=1
        else:
            continue
    
    result=1
    for i in range(2,ngt):
        if ngt%i==0 and dic1.get(i,0)!=0:
            result=i
    
    return result

n1=int(input("Enter first number:"))
n2=int(input("Enter second number:"))

print(GCD(n1,n2))
```

```python
#Least Common Multiple

def LCM(n1,n2):
    if n1==n2:
        return n1
    greater=max(n1,n2)
    lower=min(n1,n2)
    i=1
    temp=greater
    while True:
        if greater%lower==0:
            return greater
        i+=1;
        greater=temp*i
        #print(greater)
    return 0

print("LCM is :",LCM(67,47))
```
