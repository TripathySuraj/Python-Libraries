```python
#distplot by seaborn
#plots with bar graph and Kernel Destiny Estimate

#Kernel Density Estimate (KDE) is a non-parametric way 
#to estimate the probability density function (PDF) of a continuous random variable.

#KDE lets you smooth out a histogram and gives you a curve that represents the distribution of data.
```

```python
df=pd.read_csv(r'C:\Users\hp\Downloads\hr_data.csv')
sns.distplot(df["time_spend_company"])
#there is bar and curve is KDE
```

```python
sns.distplot(df["left"])
#this shows density of 0 is more so more employees have left the company
```

```python
sns.distplot(df["average_montly_hours"])
```

```python
df.describe()
```

```python
#bins
#here max time spend is 10 and min time spend is 2 so bin can be b/w 2 and 10
bins=np.arange(2,11)
print(bins)
sns.distplot(df["time_spend_company"],bins=bins)
plt.xticks(bins)
#max no of employee are there who are working between 3-4 hours
```

```python
#KDE-false - kde curve is remove, only histogram bars are kept
bins=np.arange(2,11)
sns.distplot(df["time_spend_company"],bins=bins,kde=False)
plt.xticks(bins)
```

```python
#hist -false - histogram bars are removed, only kde curve are kept
bins=np.arange(2,11)
sns.distplot(df["time_spend_company"],bins=bins,hist=False)
plt.xticks(bins)
```

```python
#rug is used to get marked hidden point
sns.distplot(df["time_spend_company"],bins=bins,rug=True)
```

```python
#hist_kws,kde_kws - kws is keyword arguments
#hist_kws is for properties of histogram
#kde_kws is for the properties of kde curve
#alpha is for transparency for the histograms
#due to apha red color of hist is slight light

sns.distplot(df["time_spend_company"],bins=bins,rug=True,hist_kws={"color":"red","edgecolor":"yellow","linewidth":3,"alpha":0.5}
            ,kde_kws={"color":"purple","linewidth":3})
plt.xticks(bins)
```

```python
#for changing both hist_kws and kde_kws at same time
sns.distplot(df["time_spend_company"],bins=bins,rug=True,color="green")
plt.xticks(bins)
```

```python
#vertical directional graph (opposite directionalgraph)
sns.distplot(df["time_spend_company"],bins=bins,rug=True,color="green",vertical=True)
plt.yticks(bins)
```

```python
#Scatter plot by seaborn
df_scatter=sns.load_dataset("titanic")
df_scatter.head()
```

```python
sns.scatterplot(x="age",y="fare",data=df_scatter)
```

```python
#hue
sns.scatterplot(x="age",y="fare",data=df_scatter,hue="alive")
#this shows most people from class having fare b/w 0 to 100 were dead
#rich people survived more
```

```python
#style and palette
plt.figure(figsize=(12,6))
sns.scatterplot(x="age",y="fare",data=df_scatter,hue="alive",style="class",palette="flare")
plt.title("Titanic Data Analysis")

#you can see maximum survivors were the people from first class represented by circles 
```

```python
#alfa
plt.figure(figsize=(12,6))
sns.scatterplot(x="age",y="fare",data=df_scatter,hue="alive",style="class",palette="flare",alpha=0.5)
plt.title("Titanic Data Analysis")
```
