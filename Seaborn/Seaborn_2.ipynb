```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#bar plot by seaborn
#bar plot represents an estimate of central tedency for a numeric variable with the height of each rectangle
#and provides some indication of the uncertainity around that estimate using error bars

df=sns.load_dataset("titanic")
df.head()
```

```python
sns.barplot(x="class",y="fare",data=df)
#the line in each rectangular bar provides some indication of the uncertainity around that estimate using error bars.
```

```python
#hue and palette
sns.barplot(x="class",y="fare",data=df,hue="sex",palette="inferno")
#here you can see females have paid more in comparision to male
#this extra info we get from hue 
#hue helps to visualize the graph according to column defined inside it
```

```python
#orient - it is for orientation 
#For horizontal orient (h) x should always be numerical
#For vertical orient (v) y should always be numerical
sns.barplot(y="class",x="fare",data=df,hue="sex",palette="inferno",orient='h') 
```

```python
#estimator
#it gives uncertainity line according to value fixed inside it
#by default uncertainity line is according mean

sns.barplot(x="class",y="fare",data=df,hue="sex",palette="inferno",estimator=np.max)
```

```python
#confidence level
#it is between 0 to 100
#it is the uncertainity line
#by ci,errcolor and errwidth we change its properties like size(height),color and width respctively

sns.barplot(x="class",y="fare",data=df,hue="sex",palette="inferno",ci=80,errcolor="green",errwidth=5)

#if ci decreses height of line decreases and if ci is set >100 it will throw error as it can be between 0-100
```

```python
#saturation
#it is betwwen 0 to 1 
#it control brightness of the plot
#0 means black and white

sns.barplot(x="class",y="fare",data=df,hue="sex",palette="inferno",ci=80,errcolor="green",errwidth=5,saturation=0.3)
```

```python
#Heat Map by seaborn

flight_df=sns.load_dataset("flights")
#sns.get_dataset_names()
flight_df.head()
```

```python
#pivot table by pandas
# Sample DataFrame
data = {
    'Department': ['Sales', 'Sales', 'HR', 'HR', 'IT', 'IT'],
    'Month': ['Jan', 'Feb', 'Jan', 'Feb', 'Jan', 'Feb'],
    'Revenue': [200, 220, 150, 160, 300, 310]
}

#index - rows, columns - cols, values - col to be aggregated on, aggfunc - parameter of aggregator like sum, max,etc

df = pd.DataFrame(data)

# Pivot Table
pivot = df.pivot_table(index='Department', columns='Month', values='Revenue', aggfunc='sum')

pivot
```

```python
#another way for pivot table here aggfunc is by default sum
flight_df_pivot=flight_df.pivot("month","year","passengers")
flight_df_pivot.head()
```

```python
plt.figure(figsize=(12,6))
ax=sns.heatmap(flight_df_pivot)
ax

#the heat map is showing lighter the color larger the value and darker the color smaller the value
#the heat map is showing the number of customers are increasing year by year
```

```python
#annot - annotation  and fmt = 'd' is for double data type value of cells
#it gives value of each block
#when background color is dark font color of values will be in white and in case of light background it will be black

plt.figure(figsize=(12,6))
ax=sns.heatmap(flight_df_pivot,annot=True,fmt='d')
ax

# Common values for fmt:
# fmt Value	Meaning	Example Output
# 'd'	Integer format	123
# '.2f'	Float with 2 decimal places	123.45
# 'g'	General format (int/float adaptive)	1e3, 123
# 's'	String format (for non-numeric data)	abc

```

```python
#linecolor and linewidth - border setting
plt.figure(figsize=(12,6))
ax=sns.heatmap(flight_df_pivot,annot=True,fmt='d',linecolor="blue",linewidth=5)  #linecolor='k' is for black color
#'b' for blue color and you can also give hexadecimal codes of color
ax
```

```python
#cmap
plt.figure(figsize=(12,6))
ax=sns.heatmap(flight_df_pivot,annot=True,fmt='d',linecolor="blue",linewidth=5,cmap="anything")
ax
```

```python
#cmap
plt.figure(figsize=(12,6))
ax=sns.heatmap(flight_df_pivot,annot=True,fmt='d',linecolor="blue",linewidth=5,cmap="PiYG")
ax

#here cbar is changed greener the value bigger the value and more the maroon less the value
```

```python
#cbar position
plt.figure(figsize=(12,6))
#to remove cbar
ax=sns.heatmap(flight_df_pivot,annot=True,fmt='d',linecolor="blue",linewidth=5,cmap="PiYG",cbar=False)
ax
```

```python
#Pair Plot
#pair plot gives various plots at the same figure
df=sns.load_dataset("tips")
df.head()
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue="sex")
plt.show()
```

```python
#vars here allows us to show plots according to particular columns only
#hue_order specifies the order of hue
#pallete specifies color of plot
sns.pairplot(df,vars=["total_bill","tip"],hue="sex",hue_order=["Female","Male"],palette="flare")
plt.show()
```

```python
#to specify columns in x-axis and y-axis
#done by x_var and y_var
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],x_vars=["total_bill","tip"],y_vars=["smoker","size","tip"])
plt.show()
```

```python
#Changing graphs in the figure
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],kind="reg") #kind can be 'scatter','kde','hist','reg'
plt.show()

# reg stands for regression. It draws a scatter plot plus a linear regression line in each subplot.
# The regression line helps to visualize the trend or linear relationship between the variables.

# kind	Description
# 'scatter'	(Default) Basic scatter plot
# 'reg'	Scatter plot with linear regression line (includes confidence band)
# 'kde'	Kernel Density Estimate (smoothed distribution) plots
# 'hist'	2D histogram for joint distributions
```

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],kind="kde") #kind can be 'scatter','kde','hist','reg'
plt.show()
```

```python
#in diagnam only kde is appearing
#to change this to hist by using diag_kind
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],kind="kde",diag_kind="hist") #kind can be 'scatter','kde','hist','reg'
plt.show()
```

```python
#to change hue marker
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],markers=['*','<']) 
plt.show()
```

```python
#to change hue marker
#for single marker
sns.pairplot(df,hue="sex",markers='*') 
plt.show()
```

```python
Main Uses of a Box Plot
✅ 1. Identify Outliers
Outliers are values that are significantly different from others in the dataset.

Box plot shows these as individual points beyond the whiskers.

✅ 2. Show Spread of the Data
It gives a clear view of how data is distributed, whether it's tight, spread out, or skewed.

✅ 3. Visualize the Median and Quartiles
Helps quickly find:

Median (Q2) – middle value

Lower quartile (Q1) – 25th percentile

Upper quartile (Q3) – 75th percentile

IQR (Interquartile Range) = Q3 - Q1

✅ 4. Compare Multiple Distributions
Useful when comparing several groups side-by-side.

Example: comparing salaries across departments or test scores across schools.

📦 Box Plot Structure Explained
markdown
Copy
Edit
             Outliers
                 ●
        ┌───────────────┐
        │               │
  Whiskers           Whiskers
    |                   |
    |     ┌───────┐     |
    |     │       │     |
    └────▶│  Q1   │     └────▶ Q3 (75th percentile)
          │       │
          └───────┘
               ▲
            Median (Q2)
Box: from Q1 (25%) to Q3 (75%) — shows the middle 50% of the data

Line inside box: the median (Q2)

Whiskers: extend to min/max values excluding outliers

Dots: outliers (points outside 1.5×IQR from the quartiles)

📊 Example (Python/Seaborn):
python
Copy
Edit
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x='department', y='salary', data=df)
plt.show()

✅ Summary
Feature	Meaning
Median line	Central tendency
Box edges	Q1 (25%) and Q3 (75%)
IQR	Middle 50% spread
Whiskers	Data range without outliers
Outlier points	Values significantly above/below normal
```

```python
df=sns.load_dataset("tips")
print(df.shape)
df.head()
```

```python
sns.set(style="whitegrid")

sns.boxplot(x="day",y="total_bill",data=df)
plt.show()
```

```python
#hue
sns.set(style="whitegrid") #background of plot

sns.boxplot(x="day",y="total_bill",data=df,hue="time",color="green")
plt.show()
```

```python
sns.set(style="darkgrid")

sns.boxplot(x="day",y="total_bill",data=df,hue="sex",color="yellow")
plt.show()
```

```python
#order in x-axis
sns.set(style="whitegrid")

sns.boxplot(x="day",y="total_bill",data=df,hue="time",color="green",order=["Fri","Sun","Thur","Sat"])
plt.show()
```

```python
#getting mean value by arrows by showmean
#also changing properties of mean arrows by meanprops
sns.set(style="whitegrid")

sns.boxplot(x="day",y="total_bill",data=df,color="red",order=["Fri","Sun","Thur","Sat"],
           showmeans=True,meanprops={"marker":'+',"markeredgecolor":"yellow","markersize":14})
plt.show()
```

```python
#markeredgecolor - for border of marker
#markerfacecolor - for inner space in marker
sns.boxplot(x="day",y="total_bill",data=df,color="red",order=["Fri","Sun","Thur","Sat"],
           showmeans=True,meanprops={"marker":'o',"markeredgecolor":"yellow","markerfacecolor":"green","markersize":14})
plt.show()
```

```python
#linewidth for lines in box plot
#palette
sns.boxplot(x="day",y="total_bill",data=df,color="red",order=["Fri","Sun","Thur","Sat"],
           showmeans=True,meanprops={"marker":'o',"markeredgecolor":"yellow","markerfacecolor":"green","markersize":14},
           linewidth=4,palette="plasma")
plt.show()
```

```python
#horizontal boxplot - orient='v' and in x-axis numeric data should be there

#first just by interchanging x and y values (orient=h optional)
#good way
sns.boxplot(x="total_bill",y="day",data=df,color="red",order=["Fri","Sun","Thur","Sat"],
           showmeans=True,meanprops={"marker":'o',"markeredgecolor":"yellow","markerfacecolor":"green","markersize":14},
           linewidth=4,palette="plasma",orient='h') #here orient is optional without orient also it will work jst y interchange
plt.show()
```

```python
#second by using orient=h (must)
#not a good way
sns.boxplot(data=df,color="red",
           showmeans=True,meanprops={"marker":'o',"markeredgecolor":"yellow","markerfacecolor":"green","markersize":10},
           linewidth=2,palette="plasma",orient='h')
plt.show()
```

```python
#single box
sns.set(style="darkgrid")

#horizontal
sns.boxplot(x=df["total_bill"])
plt.show()
```

```python
#vertical
sns.boxplot(y=df["total_bill"])
plt.show()
```

```python
#Strip Plot

#Strip plot is basically a scatter plot that differentiates different categories. It is a good alternative to box & violin plot
```

```python
df=sns.load_dataset("tips")
df.head()
```

```python
sns.set(style="whitegrid")

sns.stripplot(x="day",y="total_bill",data=df)
```

```python
#hue,palette
#linewidth is for border width of marker 
#edgecolor is for border color
#size is for size of markers

sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",edgecolor="yellow",linewidth=1,size=8)
```

```python
#jitter - displacement or space between scattered markers
#jitter=1 is original 

plt.figure(figsize=(8,8))
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",edgecolor="yellow",linewidth=1,jitter=5,size=10)
plt.show()
```

```python
#marker is for marker type
#we can also set different marker types for different hue elements

plt.figure(figsize=(5,5))
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",edgecolor="yellow",linewidth=1,jitter=5,size=10,
             marker='>')
plt.show()
```

```python
#setting marker types for different hue elements
#alpha for transparency

m={"Male":'o',"Female":'>'}
plt.figure(figsize=(5,5))
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",edgecolor="yellow",linewidth=1,jitter=5,size=10,
             marker='*',alpha=0.5)
plt.show()
```

```python
#strip plot for total bills for all days as a single unit
#for horizontal use x and for vertical plot use y

sns.stripplot(y=df["total_bill"])
```

```python
#Cat Plot

Plots that can be included in cat plots
1) Categorical scatterplots - strpplot (kind="strip"), swarmplot (kind="swarm")
2) Categorical estimate plots - pointplot (kind="point"), barplot (kind="bar"), countplot (kind="count")
3)Categorical distribution plots - boxplot (kind="box"), violinplot (kind="violin"), boxenplot (kind="boxen")
```

```python
df=sns.load_dataset("tips").head(20)
df
```

```python
#height to control height of plot
#by default scatter plot is shown in cat plot

sns.catplot(x="tip",y="size",data=df,hue="sex",palette="Oranges",height=5)
plt.show()
```

```python
#bar plot in cat plot
sns.catplot(x="day",y="size",data=df,hue="sex",palette="Oranges",kind="bar")
plt.show()
```

```python
#point plot in cat plot
sns.catplot(x="day",y="size",data=df,hue="sex",palette="Oranges",kind="point")
plt.title("Point Plot")
plt.show()

#Box plot in cat plot
sns.catplot(x="day",y="size",data=df,hue="sex",palette="Oranges",kind="box")
plt.title("Box Plot")
plt.show()

#Boxen plot in cat plot
sns.catplot(x="day",y="size",data=df,hue="sex",palette="Oranges",kind="boxen")
plt.title("Boxen Plot")
plt.show()

```
