```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Count Plot

#Difference b/w Count plots and Bar Plots
=> Countplot plots the count of the number of records by category
=> Barplot plots a value or metric for each category (by default, barplot plots the mean of variable, by category)

#Example
customers came in the restaurant all day (breakfast,lunch,dinner)
out of them how many are males and females, this overall counting can be done by countplots

while barplot gives mean data
```

```python
df=sns.load_dataset("tips")
df.head()
```

```python
sns.countplot(x="sex",data=df)
plt.title("Count of total Male-Female over all days")
plt.show()
```

```python
sns.barplot(x="sex",y="size",data=df)
plt.title("Using bar plot we get mean of size of customer for males and females")
plt.show()
# df1=df["sex"].value_counts()
# list(df1)
```

```python
#hue
sns.countplot(x="sex",data=df,hue="smoker")
plt.title("Count of total Male-Female over all days")
plt.show()
```

```python
#horizontal plot
sns.countplot(y="sex",data=df,hue="smoker")
plt.title("Count of total Male-Female over all days")
plt.show()
```

```python
#palette
sns.countplot(x="sex",data=df,hue="smoker",palette="flare")
plt.title("Count of total Male-Female over all days")
plt.show()
```

```python
#color and alpha
sns.countplot(x="sex",data=df,hue="smoker",color="yellow",alpha=0.5)
plt.title("Count of total Male-Female over all days")
plt.show()
```

```python
#Point plot

# A point plot in Seaborn (sns.pointplot()) is used to visualize the mean (or another estimator) of a numerical variable 
# for one or more categorical variables, along with confidence intervals.

dft=sns.load_dataset("titanic")
dft
```

```python
sns.set(style="darkgrid")
plt.figure(figsize=(8,6))

sns.pointplot(x="class",y="age",data=dft)
plt.plot()

#this gives mean age of travellers belonging to different classes

#the three vertical lines passing through points are confidence intervals (CI) 

#default CI is 95%
# A 95% confidence interval (CI) means:
# "We are 95% confident that the true population parameter (like the mean or proportion) lies within this interval."
#Suppose:
# You survey 100 people and find the average height is 170 cm
# You calculate a 95% confidence interval: [168.5, 171.5]
# ✅ Interpretation:
# We are 95% confident that the true average height of the entire population lies between 168.5 cm and 171.5 cm.
```

```python
#changing estimator to median (instead of mean which is by default)

sns.pointplot(x="class",y="age",data=dft,estimator=np.median)
plt.plot()
```

```python
#changing CI
sns.pointplot(x="class",y="age",data=dft,ci=80)
plt.plot()
```

```python
#for standard deviation put it in ci instead of estimator
sns.pointplot(x="class",y="age",data=dft,ci="sd")
plt.plot()
```

```python

sns.pointplot(x="class",y="age",data=dft,hue="sex",palette=["red","green"],markers=['o','x'],linestyles=['-','--']
             ,order=["Third","First","Second"])
plt.plot()
```

```python
#adding cap to CI lines
#removing joining lines
sns.pointplot(x="class",y="age",data=dft,capsize=0.1,join=False)
plt.plot()
```

```python
#combining striplot and pointplot
plt.figure(figsize=(8,6))

sns.stripplot(x="class",y="age",data=dft,alpha=0.5)
sns.pointplot(x="class",y="age",data=dft,join=False,color="red")
plt.show()

#here you can see where the density of points in scatter plot is high, mean value is also there in point plot
```

```python
#combining boxplot and pointplot
plt.figure(figsize=(8,6))

sns.boxplot(x="class",y="age",data=dft)
sns.pointplot(x="class",y="age",data=dft,join=False,color="red")
plt.show()

#here you can see where the mean (Q1) in scatter plot is there, mean value is also there in point plot
```

```python
#using columns to show plot according to a particular one 
#similar to hue but showed in different plots
#here catplot will be used
#aspect is width/height ratio

sns.catplot(x="class",y="age",data=dft,kind="point",col="embark_town",height=5,aspect=0.8,hue="alive")
plt.plot()
```

```python
#Multiple plots in same graph in seaborn
#also called Facet-Grid 
#suppose you want graph of both male and female data in
#seperate graphs then Facet-Grid is used
#like hue attribute but in seperate plots (for understanding)
df=sns.load_dataset("tips")
df.head()
```

```python
fg=sns.FacetGrid(df)

#scatter plot is used so inside map it is set
#then x and y values of scatter plot is put
#here matplotlib scatter plot is used

fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

```python
#for multiplot use col

fg=sns.FacetGrid(df,col="sex")
fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

```python
#using hue and adding legend
fg=sns.FacetGrid(df,col="sex",hue="day")
fg.map(plt.scatter,"total_bill","tip").add_legend()
plt.show()
```

```python
#using bar plot
plt.figure(figsize=(12,12))
fg=sns.FacetGrid(df,col="sex",hue="day",palette="flare")
fg.map(plt.bar,"total_bill","tip",edgecolor="yellow").add_legend()
plt.show()
```

```python
#Using new dataset
pen=sns.load_dataset("penguins")
pen.head()
```

```python
#by .map
g=sns.FacetGrid(pen,col="island")
g.map(sns.histplot,"flipper_length_mm")
```

```python
#by .map_dataframe
#here x and y can be given so horizontal and vertical plots can be made
g=sns.FacetGrid(pen,col="island")
g.map_dataframe(sns.histplot,x="flipper_length_mm")
```

```python
g=sns.FacetGrid(pen,col="island")
g.map_dataframe(sns.histplot,y="flipper_length_mm")
```

```python
#sharey=False will make y axis seperate for each plot
#ylim will make limit in each y axis same

g=sns.FacetGrid(pen,col="island",sharey=False,ylim=(20,70),hue="species",palette="prism")
g.map_dataframe(sns.scatterplot,x="bill_depth_mm",y="bill_length_mm")
g.add_legend()
g.set_axis_labels("bill Depth (mm)","bill Length (mm)")
g.set_titles(col_template="{col_name} island")
```

```python
#adding both row and column

quantity="body_mass_g"

g=sns.FacetGrid(pen,row="sex",col="species",sharey=False,palette="prism")
g.map_dataframe(sns.kdeplot,x=quantity,linewidth=4,shade=True)

g.set_titles(row_template="{row_name}",col_template="{col_name} Penguins")
```

```python
#scatterplot
m={"Male":"*","Female":">"}
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=pen,hue="sex",style="sex",size="sex",sizes=(200,100),
               palette="flare",alpha=1,markers=m)

plt.show()
```

```python
#Violin Plot
#similar to box plot

df=sns.load_dataset("tips")
df.head()
```

```python
#horizontal
sns.violinplot(y="day",x="total_bill",data=df,palette="flare",hue="sex")
plt.show()
```

```python
#inner
sns.violinplot(x="time",y="total_bill",data=df,linewidth=3,order=["Dinner","Lunch"],palette="flare",inner="quart")
#inner can have quart,point,box,stick and None
plt.show()
```

```python
sns.violinplot(x="time",y="total_bill",data=df,linewidth=3,order=["Dinner","Lunch"],palette="flare",inner="box") #box is default
plt.show()
```

```python
sns.violinplot(x="time",y="total_bill",data=df,linewidth=3,order=["Dinner","Lunch"],palette="flare",inner="stick")
plt.show()
```

```python
#single violin plot
#take only columns having numerical values

sns.violinplot(y=df["total_bill"]) #use x for horizontal
plt.show()
```

```python
#swarmplot
#here point dont overlap like in case of stripplot
#here first sort all points then plot them
#it appears like a swarm of bees thats why its name is swarmplot

cars=sns.load_dataset('mpg').dropna()
print(cars.shape)
cars.head()
```

```python
cars.cylinders.value_counts()
```

```python
#cars with even number of cylinders
c=pd.DataFrame(cars.cylinders.value_counts())
print(c.index)
print(c["cylinders"].values)
```

```python
cars=cars[cars.cylinders.isin([4,6,8])]
cars.head()
```

```python
sns.set(style="whitegrid")
sns.swarmplot(cars.horsepower)
```

```python
#size is the size of dots
#setting x and y here
#linewidth - border width
#edgecolor - border color
#it shows all properties as shown in scatterplot by matplotlib

sns.swarmplot(x=cars.origin,y=cars.horsepower,size=10,linewidth=1,palette="BuPu",edgecolor="orange",marker="*")
plt.show()
```

```python
#hue
sns.swarmplot(x="origin",y="horsepower",data=cars,hue="cylinders")
plt.show()
```

```python
#splitting each group in hue horizontally by using dodge which False by default

sns.swarmplot(x="origin",y="horsepower",data=cars,hue="cylinders",dodge=True)
plt.show()
```

```python
usa=cars[cars["origin"]=="usa"]
usa.head()
```

```python
#boxplot with swarmplot
#whis=np.inf puts whispers to the extent data goes

sns.boxplot(x="cylinders",y="horsepower",data=usa,color="yellow",whis=np.inf)
sns.swarmplot(x="cylinders",y="horsepower",data=usa,alpha=0.7)
```

```python
#violinplot with swarmplot

sns.violinplot(x="cylinders",y="horsepower",data=usa,scale="width",inner=None,color="darkgreen")
sns.swarmplot(x="cylinders",y="horsepower",data=usa,alpha=0.7,size=4,color="white")
```

```python
#as you can see from above plots points lie within the box and whiskers (max-min limit) 
#in violin - swarmplot all points are within the violin shaped plot
#swarmplot shows scatterplot categorical distribution like the stripplot
#in stripplot points may overlap but in swarmplot points do not overlap
```
